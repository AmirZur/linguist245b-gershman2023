```python
import os
import pandas as pd

data = []
for fn in os.listdir('responses/pilot_b'):
    participant_df = pd.read_csv(f'responses/pilot_b/{fn}')
    participant_df['participant_id'] = fn.split('.')[0]
    data.append(participant_df)

df = pd.concat(data)

df.to_csv('pilot_b.csv')
```

```python
import plotly.express as px

condition = {
    'prob_lower': 'X is associated with a lower probability of Y',
    'risk_lower': 'X is associated with a lower risk of Y',
    'prob_increase': 'X is associated with an increased probability of Y',
    'risk_increase': 'X is associated with an increased risk of Y',
    'assoc': 'X is associated with Y',
}

plot_df = df[df['task'] == 'causal_judgment'].copy()
plot_df['condition'] = plot_df['condition'].map(condition)

plot_df['response'] = plot_df['response_type'].map({
    'xy': 'X -> Y',
    'yx': 'Y -> X',
    'neither': 'neither'
})

px.histogram(
    plot_df,
    x='response',
    facet_col='condition',
    template='simple_white'
)
```
